# Part-of-speech (POS)

Part-of-speech (POS) tagging is a popular NLP technique which refers to categorizing words in a text (corpus) in correspondence with a particular part of speech, depending on the definition of the word and its context.

### pos_tag method

````python
import re

# Offline helpers: the lab can run even when NLTK data cannot be downloaded.
def simple_tokenize(text):
    return re.findall(r"[A-Za-z]+(?:'[A-Za-z]+)?|[0-9]+|[.,!?;:]", text)

def simple_sent_tokenize(text):
    return [part.strip() for part in re.split(r"(?<=[.!?])\s+", text.strip()) if part.strip()]

POS_RULES = {
    'i': 'PRP', 'you': 'PRP', 'he': 'PRP', 'she': 'PRP', 'it': 'PRP',
    'we': 'PRP', 'they': 'PRP', 'my': 'PRP$', 'your': 'PRP$',
    'the': 'DT', 'a': 'DT', 'an': 'DT', 'this': 'DT',
    'is': 'VBZ', 'am': 'VBP', 'are': 'VBP', 'was': 'VBD', 'were': 'VBD',
    'do': 'VBP', 'does': 'VBZ', 'did': 'VBD', 'to': 'TO',
    'in': 'IN', 'on': 'IN', 'of': 'IN', 'from': 'IN', 'and': 'CC', 'or': 'CC',
    'not': 'RB', 'very': 'RB', 'how': 'WRB',
    'play': 'VB', 'watching': 'VBG', 'received': 'VBD', 'broken': 'VBN',
    'like': 'VBP', 'hated': 'VBD', 'search': 'VB', 'searched': 'VBD',
    'google': 'VB', 'table': 'VB', 'football': 'NN', 'matches': 'NNS',
    'phone': 'NN', 'amazon': 'NNP', 'lecture': 'NN', 'movie': 'NN',
    'transformers': 'NNS', 'childhood': 'NN', 'food': 'NN', 'motion': 'NN'
}

def simple_pos_tag(tokens):
    tagged = []
    for token in tokens:
        lower = token.lower()
        if token in {'.', ',', '!', '?', ';', ':'}:
            tag = token
        elif lower in POS_RULES:
            tag = POS_RULES[lower]
        elif lower.endswith('ing'):
            tag = 'VBG'
        elif lower.endswith('ed'):
            tag = 'VBD'
        elif lower.endswith('ly'):
            tag = 'RB'
        elif lower.endswith('s'):
            tag = 'NNS'
        else:
            tag = 'NN'
        tagged.append((token, tag))
    return tagged
````

````python
text = "I received my new phone broken from Amazon."
tokens = simple_tokenize(text)

for word, tag in simple_pos_tag(tokens):
    print(f'word: ({word}), part of speech: ({tag})')
````

````python
try:
    import spacy
    nlp = spacy.load('en_core_web_sm')
    SPACY_AVAILABLE = True
except Exception:
    SPACY_AVAILABLE = False
    print('spaCy model is unavailable; using a simple offline POS fallback.')
    POS_MAP = {
        'NN': 'NOUN', 'NNS': 'NOUN', 'NNP': 'PROPN', 'NNPS': 'PROPN',
        'JJ': 'ADJ', 'JJR': 'ADJ', 'JJS': 'ADJ',
        'RB': 'ADV', 'RBR': 'ADV', 'RBS': 'ADV',
        'VB': 'VERB', 'VBD': 'VERB', 'VBG': 'VERB', 'VBN': 'VERB', 'VBP': 'VERB', 'VBZ': 'VERB',
        'PRP': 'PRON', 'PRP$': 'PRON', 'DT': 'DET', 'IN': 'ADP',
        'CC': 'CCONJ', 'CD': 'NUM', 'MD': 'AUX', 'TO': 'PART', 'RP': 'PART', 'UH': 'INTJ'
    }
    TAG_EXPLANATIONS = {
        'NN': 'noun, singular or mass', 'NNS': 'noun, plural', 'NNP': 'proper noun, singular',
        'JJ': 'adjective', 'RB': 'adverb', 'VB': 'verb, base form', 'VBD': 'verb, past tense',
        'VBG': 'verb, gerund or present participle', 'VBN': 'verb, past participle',
        'PRP': 'personal pronoun', 'DT': 'determiner', 'IN': 'preposition or conjunction'
    }
    class FallbackToken:
        def __init__(self, text, tag):
            self.text = text
            self.tag_ = tag
            self.pos_ = POS_MAP.get(tag, 'PUNCT' if tag in {'.', ',', ':', '``', "''", '-LRB-', '-RRB-'} else 'X')
    class FallbackDoc(list):
        pass
    class FallbackNLP:
        def __call__(self, text):
            words = simple_tokenize(text)
            return FallbackDoc([FallbackToken(word, tag) for word, tag in simple_pos_tag(words)])
    class FallbackSpacy:
        @staticmethod
        def explain(tag):
            return TAG_EXPLANATIONS.get(tag, tag)
    class FallbackDisplacy:
        @staticmethod
        def render(doc, style='dep', jupyter=True, options=None):
            print('Dependency visualization is unavailable without spaCy; tokens and POS tags:')
            print(' | '.join(f'{token.text}/{token.pos_}' for token in doc))
    nlp = FallbackNLP()
    spacy = FallbackSpacy()
    displacy = FallbackDisplacy()
````

````python
doc = nlp(u"The quick brown fox jumped over the lazy dog's back.")
````

* To view the coarse POS tag use `token.pos_`
* To view the fine-grained tag use `token.tag_`
* To view the description of either type of tag use `spacy.explain(tag)`

````python
for token in doc:
    print(f'{token.text:{10}} {token.pos_:{8}} {token.tag_:{6}} {spacy.explain(token.tag_)}')
````

In the English language, the same word can have different meanings, even within the same sentence.

## Example: Find the adjective words in a sentance


````python
# Reuse the offline tokenization and POS helpers from the setup cell.
````

````python
text =  "Mobile is very handy. Packed with good performances and nice camera. This is my first iPhone. Very good experience with Apple. Battery is worst. I am hardly getting 5 hours screen on time. Positive thing is I came to know how many hours I was spending on my mobile. Because of poor battery now I am only using mobile for 4 hours a day."

print(text, end='\n\n')

for sent in simple_sent_tokenize(text):
    tokens = simple_tokenize(sent)
    part_of_speech = simple_pos_tag(tokens)

    for i in part_of_speech:
        if 'JJ' in i:
            print(i)
    print('============================')
````
